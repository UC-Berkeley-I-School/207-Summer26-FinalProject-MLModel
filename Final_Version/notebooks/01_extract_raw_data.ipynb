# Extract Raw Data

Official **BTS**, **NOAA**, and **FAA** files are reused when valid and downloaded only when missing. 

### Data Pull

- **BTS:** all flights touching SFO (departures + arrivals)
- **NOAA:** only the SFO hourly station, one parquet file per calendar year
- **FAA:** the releasable active and deregistered registries (one row per tail)

```python
from pathlib import Path
import json
import sys

SEARCH_ROOT = Path.cwd().resolve()
for candidate in [SEARCH_ROOT, *SEARCH_ROOT.parents]:
    site_packages = candidate / ".venv" / "Lib" / "site-packages"
    if site_packages.exists() and str(site_packages) not in sys.path:
        sys.path.insert(0, str(site_packages))
        break

ROOT = SEARCH_ROOT
if not (ROOT / "data").exists():
    ROOT = ROOT.parent

if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

CONFIG_PATH = ROOT / "config" / "project_config.json"
CONFIG = json.loads(CONFIG_PATH.read_text(encoding="utf-8"))
```

```python
from src.data_pull import FlightDataPull

pull = FlightDataPull(
    ROOT,
    airport=CONFIG["airport"],
    start_date=CONFIG["start_date"],
    end_date_exclusive=CONFIG["end_date_exclusive"],
    noaa_station_id=CONFIG["noaa_station_id"],
    tracked_airlines=tuple(CONFIG["airlines"]),
)

source_status = pull.ensure_all(allow_download=True, deep=True)
display(source_status)
```
